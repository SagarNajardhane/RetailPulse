```python
# ============================================
# RETAILPULSE - DATA EXPLORATION (Part 1)
# Loading and First Look at the Dataset
# ============================================

# Import the libraries we need
import pandas as pd        # For loading and manipulating data (like Excel for Python)
import numpy as np         # For mathematical operations
import os                  # For working with file paths
import warnings
warnings.filterwarnings('ignore')  # Hide unnecessary warning messages

print("✅ Libraries imported successfully!")
```

```python
# ============================================
# LOAD THE DATASET
# ============================================

# Define the path to our data files
# We use os.path to make sure the path works on any computer (Windows/Mac/Linux)
data_dir = os.path.join("..", "data", "raw")

file_2009 = os.path.join(data_dir, "Year 2009-2010.xlsx")
file_2010 = os.path.join(data_dir, "Year 2010-2011.xlsx")

print("📂 Looking for data files...")
print(f"   File 1: {file_2009}")
print(f"   Exists: {os.path.exists(file_2009)}")
print(f"   File 2: {file_2010}")
print(f"   Exists: {os.path.exists(file_2010)}")
print()

# Load the first file (2009-2010)
print("⏳ Loading Year 2009-2010 data... (this may take 30-60 seconds)")
df_2009 = pd.read_excel(file_2009)
print(f"   ✅ Loaded {len(df_2009):,} rows")

# Load the second file (2010-2011)
print("⏳ Loading Year 2010-2011 data... (this may take 30-60 seconds)")
df_2010 = pd.read_excel(file_2010)
print(f"   ✅ Loaded {len(df_2010):,} rows")

# Combine both files into one big DataFrame
print("\n🔗 Combining both datasets...")
df = pd.concat([df_2009, df_2010], ignore_index=True)
print(f"   ✅ Combined dataset has {len(df):,} rows and {len(df.columns)} columns")

# Free up memory by deleting the individual DataFrames
del df_2009, df_2010

print("\n🎉 Dataset loaded successfully!")
```

```python
# ============================================
# FIRST LOOK AT THE DATA
# ============================================

# .head() shows the first 5 rows of the DataFrame
print("📊 FIRST 5 ROWS OF THE DATASET:")
print("=" * 80)
df.head()
```

```python
# ============================================
# UNDERSTAND THE COLUMNS
# ============================================

print("📋 COLUMN DESCRIPTIONS:")
print("=" * 80)
print("""
Invoice     → Invoice number. A unique number for each purchase.
              If it starts with 'C', it means the order was CANCELLED (returned).

StockCode   → Product code. A unique identifier for each product.
              Example: '85048' is a specific Christmas decoration.

Description → Product name in plain English.
              Example: '15CM CHRISTMAS GLASS BALL 20 LIGHTS'

Quantity    → How many units of this product were bought.
              Negative numbers mean the item was RETURNED.

InvoiceDate → The date and time when the purchase was made.

Price       → The price per unit in British Pounds (£).

Customer ID → A unique number for each customer.
              Some rows have no Customer ID (NaN) — these are guest purchases.

Country     → The country where the customer is located.
""")

# Show the actual column names in our DataFrame
print("📌 Actual column names in our DataFrame:")
for i, col in enumerate(df.columns, 1):
    print(f"   {i}. {col}")
```

```python
# ============================================
# BASIC STATISTICS
# ============================================

print("📊 DATASET SHAPE (Rows × Columns):")
print(f"   Rows:    {df.shape[0]:,}")
print(f"   Columns: {df.shape[1]}")
print()

print("📊 DATA TYPES (What kind of data is in each column?):")
print("-" * 40)
print(df.dtypes)
print()

print("📊 MISSING VALUES (How many empty cells in each column?):")
print("-" * 40)
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct.round(2)
})
print(missing_df)
print()

print("📊 NUMERICAL STATISTICS:")
print("-" * 40)
df.describe()
```

```python
# ============================================
# QUICK SUMMARY
# ============================================

print("🏪 RETAILPULSE DATASET SUMMARY")
print("=" * 60)
print(f"📅 Date Range:     {df['InvoiceDate'].min()} to {df['InvoiceDate'].max()}")
print(f"🧾 Total Invoices: {df['Invoice'].nunique():,}")
print(f"📦 Total Products: {df['StockCode'].nunique():,}")
print(f"👥 Total Customers:{df['Customer ID'].nunique():,}")
print(f"🌍 Total Countries:{df['Country'].nunique()}")
print(f"💰 Price Range:    £{df['Price'].min():.2f} to £{df['Price'].max():.2f}")
print(f"📊 Quantity Range: {df['Quantity'].min():,} to {df['Quantity'].max():,}")
print()

# Top 5 countries
print("🌍 TOP 5 COUNTRIES BY NUMBER OF TRANSACTIONS:")
print("-" * 40)
top_countries = df['Country'].value_counts().head(5)
for country, count in top_countries.items():
    pct = (count / len(df)) * 100
    print(f"   {country:25s} {count:>8,} ({pct:.1f}%)")
print()

# Cancellations
cancellations = df[df['Invoice'].str.startswith('C', na=False)]
print(f"🔄 CANCELLED ORDERS: {len(cancellations):,} ({len(cancellations)/len(df)*100:.1f}% of all transactions)")
print()

# Returns (negative quantity)
returns = df[df['Quantity'] < 0]
print(f"↩️  RETURNED ITEMS (negative quantity): {len(returns):,} rows")
print()

# Zero or negative prices
free_items = df[df['Price'] <= 0]
print(f"🆓 FREE/NEGATIVE PRICE ITEMS: {len(free_items):,} rows")
```

```python
# ============================================
# SAVE THE COMBINED DATA
# ============================================

output_dir = os.path.join("..", "data", "processed")
os.makedirs(output_dir, exist_ok=True)

# Save as CSV (easy to open in Excel)
csv_path = os.path.join(output_dir, "online_retail_combined.csv")
print(f"💾 Saving CSV to: {csv_path}")
df.to_csv(csv_path, index=False)
csv_size = os.path.getsize(csv_path) / (1024 * 1024)
print(f"   ✅ Saved! File size: {csv_size:.1f} MB")

# Save as Parquet (fast loading for Python)
try:
    parquet_path = os.path.join(output_dir, "online_retail_combined.parquet")
    print(f"💾 Saving Parquet to: {parquet_path}")
    df.to_parquet(parquet_path, index=False)
    parquet_size = os.path.getsize(parquet_path) / (1024 * 1024)
    print(f"   ✅ Saved! File size: {parquet_size:.1f} MB")
except Exception as e:
    print(f"   ⚠️ Parquet save failed: {e}")
    print("   No worries, we'll use the CSV file instead.")

print("\n🎉 Data saved successfully!")
print("   From now on, we can load the data in seconds instead of minutes!")
```

```python
# ============================================
# BULLETPROOF PATH SETUP & DATA SAVING (WITH TYPE FIX)
# ============================================
import os
import time
import pandas as pd

# 1. Automatically detect project root
current_dir = os.getcwd()
if os.path.basename(current_dir) == "notebooks":
    project_root = os.path.abspath(os.path.join(current_dir, ".."))
else:
    project_root = os.path.abspath(current_dir)

processed_dir = os.path.join(project_root, "data", "processed")
os.makedirs(processed_dir, exist_ok=True)

csv_path = os.path.join(processed_dir, "online_retail_combined.csv")
parquet_path = os.path.join(processed_dir, "online_retail_combined.parquet")

print(f"📂 Project Root:   {project_root}")
print(f"📂 Processed Dir:  {processed_dir}")
print()

# 2. Fix mixed text data types so PyArrow doesn't get confused
print("🔧 Harmonizing column data types...")
df['Invoice'] = df['Invoice'].astype(str)
df['StockCode'] = df['StockCode'].astype(str)
df['Description'] = df['Description'].astype(str)
df['Country'] = df['Country'].astype(str)
print("   ✅ Text columns harmonized!")

# 3. Save CSV
print("\n💾 Saving CSV file...")
start = time.time()
df.to_csv(csv_path, index=False)
print(f"   ✅ Saved CSV in {time.time() - start:.2f}s (Size: {os.path.getsize(csv_path)/(1024*1024):.1f} MB)")

# 4. Save Parquet
print("\n💾 Saving Parquet file...")
start = time.time()
df.to_parquet(parquet_path, index=False, engine='pyarrow')
print(f"   ✅ Saved Parquet in {time.time() - start:.2f}s (Size: {os.path.getsize(parquet_path)/(1024*1024):.1f} MB)")

# 5. Test Fast Loading Immediately
print("\n⚡ Testing Fast Load from Parquet...")
start = time.time()
df_fast = pd.read_parquet(parquet_path)
end = time.time()

print(f"   ✅ Successfully loaded {len(df_fast):,} rows in {end - start:.2f} seconds!")
print(f"   ✅ Verification: Rows match original DataFrame? -> {len(df_fast) == len(df)}")

del df_fast  # free memory
```
